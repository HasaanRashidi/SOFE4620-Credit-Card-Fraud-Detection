# Phase 6 – Model Comparison and Final Conclusion

Course: SOFE4620 Machine Learning and Data Mining  
Project: Credit Card Fraud Detection

### Purpose of Phase 6

This phase compares the performance of all developed models and identifies which model performed best for the fraud detection task. Since the dataset is highly imbalanced, the comparison focuses on precision, recall, and F1-score rather than accuracy alone.

```python
import pandas as pd

results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Tuned Random Forest",
        "Random Forest"
    ],
    "Accuracy": [0.9991749, 0.9989291, 0.9984024, 0.9995084],
    "Precision": [0.8311688, 0.6761905, 0.5209581, 0.9605263],
    "Recall": [0.6530612, 0.7244898, 0.8877551, 0.7448980],
    "F1-score": [0.7314286, 0.6995074, 0.6566038, 0.8390805],
    "True Positives": [64, 71, 87, 73],
    "False Positives": [13, 34, 80, 3],
    "False Negatives": [34, 27, 11, 25]
})

results
```

### Comparison of Models

The Logistic Regression model served as a useful baseline and achieved good precision, but its recall was lower than the tree-based models, meaning it missed more fraudulent transactions.

The Decision Tree model improved recall compared to Logistic Regression, but this came with reduced precision and a lower F1-score.

The tuned Random Forest model achieved the highest recall, meaning it detected the largest number of fraud cases and missed the fewest fraudulent transactions. However, it also produced the most false positives, which lowered its precision.

The standard Random Forest model achieved the highest precision and the highest F1-score while still maintaining strong recall. It also produced only three false positives, giving it the best overall balance between detecting fraud and avoiding unnecessary alerts.

### Final Recommendation

Based on the evaluation results, the standard Random Forest model is the best overall model for this project. It achieved the strongest balance between precision and recall, as shown by its high F1-score.

If the priority is to minimize missed fraud cases as much as possible, then the tuned Random Forest model may be preferred because it achieved the highest recall. However, this comes at the cost of more false positives.

Overall, the Random Forest model is the most balanced and practical choice for the fraud detection task in this project.

### Limitations

This project has several limitations. First, the dataset is highly imbalanced, which makes fraud detection difficult and can affect how model performance is interpreted. Second, the dataset uses PCA-transformed features (V1 to V28), which reduces interpretability because the original feature meanings are not available. Third, the models were evaluated on one dataset and one train-test split, so performance may vary on other datasets or in real-world deployment. Finally, this project focuses on model comparison in an academic setting and is not intended to represent a complete production fraud detection system.

### Conclusion

This project successfully developed and evaluated multiple machine learning models for credit card fraud detection. The results showed that tree-based models performed strongly on this task, especially the Random Forest approaches.

Among all models, the standard Random Forest model provided the best overall performance, while the tuned Random Forest model offered the highest recall. These findings show that model choice depends on whether the priority is balanced performance or maximizing fraud detection.
