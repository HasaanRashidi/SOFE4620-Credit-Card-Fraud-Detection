```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/creditcard.csv")
df.head()
```

```python
df["Class"].value_counts()
```

```python
df["Class"].value_counts(normalize=True) * 100
```

### Visualization of Class Imbalance

```python
sns.countplot(x="Class", data=df)

plt.title("Distribution of Legitimate vs Fraudulent Transactions")
plt.xlabel("Transaction Type (0 = Legit, 1 = Fraud)")
plt.ylabel("Number of Transactions")

plt.show()
```

### Percentage Distribution of Fraudulent vs Legitimate Transactions

```python
fraud_percentage = df["Class"].value_counts(normalize=True) * 100

plt.figure(figsize=(6,6))
plt.pie(
    fraud_percentage,
    labels=["Legitimate", "Fraud"],
    autopct="%1.2f%%",
    startangle=90
)

plt.title("Percentage of Fraudulent vs Legitimate Transactions")
plt.show()
```

The dataset is highly imbalanced. Legitimate transactions represent approximately 99.8% of the data, while fraudulent transactions account for only about 0.17%.

This imbalance creates challenges for machine learning models. A model that predicts every transaction as legitimate would still achieve very high accuracy. Because of this, accuracy alone is not an appropriate metric for evaluating fraud detection models. Metrics such as precision, recall, and F1-score will be more meaningful for evaluating model performance.
